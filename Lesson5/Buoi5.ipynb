```python
import pandas as pd
df = pd.read_csv("car-sales-data.csv")
# df.drop(["Colour", "Doors"])
df["Odometer"] = df["Odometer"].fillna(df['Odometer'].mean())
df["Odometer"] = df["Odometer"].round(0).astype(int)

df['Price'] = df['Price'].fillna('$0')
df
```

```python
df.drop(["Colour", "Doors"], replace=True)
df["Odometer"] = df["Odometer"].fillna(df['Odometer'].mean())
df["Odometer"] = df["Odometer"].round(0).astype(int)

df['Price'] = df['Price'].fillna('$0')
df
```

```python
df['Price'] = df['Price'].apply(lambda x: float(x.replace('$','').replace(',','')))
df
```

```python
max_price = df['Price'].max()
max_price_car = df[df['Price'] == max_price]
max_price_car
```

```python
df_valid = df[df['Price'] > 0]
df_valid['Price_per_km'] = df_valid['Price'] / df_valid['Odometer']

min_ratio = df_valid['Price_per_km'].min()
cheapest_per_km = df_valid[df_valid['Price_per_km'] == min_ratio]
cheapest_per_km 
```

```python
average_price_by_make = df[["Make", "Price"]].groupby("Make").mean().round(0)
average_price_by_make
```

```python
max_average_price = average_price_by_make.max()
highest_avg_make = average_price_by_make[average_price_by_make == max_average_price]
highest_avg_make
```
