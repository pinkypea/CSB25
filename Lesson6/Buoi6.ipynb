```python
import pandas as pd
df = pd.read_csv("spotify.csv")
df
```

```python
# Chuyển cột Date thành kiểu dữ liệu ngày tháng năm
df["Date"] = pd.to_datetime(df["Date"], format="%Y-%m-%d")
df.dtypes
```

```python
import matplotlib.pyplot as plt
plt.figure(figsize=(14, 6)) # xác định kích thước biểu đồ
df_line = df.set_index("Date")
plt.plot(df_line) # vẽ biểu đồ đường cho toàn bộ các cột trong biểu đồ
plt.legend(df_line.columns) # hiển thị chú thích cho từng đường
plt.ylabel("Views") # gán nhãn cho trục dọc
plt.title("Spotify song views from January 2017 to January 2018") # gán tên cho biểu đồ
plt.show() # hiển thị biểu đồ
```

```python
import pandas as pd
df = pd.read_csv("2024 Billionaire List.csv", index_col=0)
df["2024 Net Worth"] = df["2024 Net Worth"].apply(lambda x : x[1 : -1])
df["2024 Net Worth"] = pd.to_numeric(df["2024 Net Worth"])
df_industry = df[["Industry", "2024 Net Worth"]].groupby("Industry").sum()
df_industry = df_industry.sort_values(by="2024 Net Worth", ascending=False)
df_top_industry = df_industry[:10]

# Tạo biểu đồ cột với trục ngang là Industry, trục dọc và Net Worth
plt.bar(x=df_top_industry.index, height=df_top_industry["2024 Net Worth"]) 
plt.xticks(rotation=45) # Xoay chữ
plt.ylabel("Net Worth ($B)")
plt.title("Top 10 Net Worth by Industry")
plt.show()
```

```python
import pandas as pd
df = pd.read_csv("2024 Billionaire List.csv", index_col=0)
df["Age"] = df["Age"].astype('Int64')

plt.hist(
    x = df["Age"],
    edgecolor = 'white'
)

plt.title("Forbes's 2024 Bilionaires Age Distribution")
plt.xlabel("Age")
plt.ylabel("Count")
plt.show()

# => Xu hướng phân bố độ tuổi của các tỷ phú trong khoảng 60 - 80 tuổi là nhiều nhất.
```

```python
import pandas as pd
df = pd.read_csv("2024 Billionaire List.csv", index_col=0)
df["2024 Net Worth"] = df["2024 Net Worth"].apply(lambda x : x[1 : -1])
df["2024 Net Worth"] = pd.to_numeric(df["2024 Net Worth"])
plt.scatter(
    x=df["Age"],
    y = df["2024 Net Worth"],
    s = df["2024 Net Worth"] * 5, 
    linewidths = 0.1,
    edgecolors = 'black'
)

plt.title("2024 Billionares's Age w.r.t Net Worth")
plt.xlabel("Age")
plt.ylabel("Net Worth ($B)")
plt.show()
```
