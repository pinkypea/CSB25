```python
import pandas as pd

data = {
    "Tên": ["An", "Bình", "Chi"],
    "Toán": [8, 6, 9],
    "Văn": [7, 8, 9],
    "Anh": [9, 5, 8]
}

df = pd.DataFrame(data)

# Lưu ra file CSV
df.to_csv("diem.csv", index=False, encoding="utf-8")
print("Đã lưu ra file diem.csv")

df.to_excel("diem.xlsx", index=False, engine="openpyxl")
print("Đã lưu ra file diem.xlsx")
```

```python
# Khởi tạo DataFrame bằng Dictionary
import pandas as pd
animals = pd.DataFrame({
    "dong_vat": ["Sao la", "Khỉ đuôi ngắn", "Voi châu Á"],
    "tuoi_tho": [25, 25, 70],
    "loai": ["có vú", "linh trưởng", "có vú"],
    "thuc_an": ["thực vật", "thực vật", "thực vật"],
    "moi_truong_song": ["Rừng", "Rừng", "Rừng"],
    "tap_tinh": ["Bầy đàn", "Bầy đàn", "Bầy đàn"]
})

animals
```

```python
# Hiển thị thông tin về bảng (số cột, số hàng, kiểu dữ liệu)
animals.info()
```

```python
animals.head(2) # Xem n dòng đầu tiên
animals.tail(2) # Xem n dòng cuối cùng
```

```python
# Dùng index slicing để lấy ra n hàng
animals[:2]
```

```python
# Lấy ra 1 Series (1 cột)
animals["dong_vat"]
```

```python
# Lấy ra nhiều cột
animals[["dong_vat", "loai"]]
```

```python
# Lọc có điều kiện
age_data = animals[["dong_vat", "tuoi_tho"]]
age_data[animals["tuoi_tho"] <= 30]
```

```python
# Thay đổi index
animals.set_index("dong_vat", inplace=True)
animals
```

```python
# Đặt index lại thành số
animals.reset_index()
```

```python
# Lọc dựa trên index
animals.iloc[:2, 0:3]
```

```python
# Lọc dựa trên dữ liệu
animals.loc[["Sao la", "Khỉ đuôi ngắn"], "tuoi_tho": "thuc_an"]
```

```python
# Lọc dữ liệu với điều kiện
animals.loc[animals["loai"] == "có vú", "tuoi_tho": "moi_truong_song"]
```
