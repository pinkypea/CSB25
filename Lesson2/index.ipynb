```python
import pandas as pd
import numpy as np
```

Chuẩn bị dữ liệu

```python
with open("temperatures.csv", "r") as file:
    df = pd.read_csv(file)
df.head()
```

Chuyển đổi input và output thành dạng numpy.ndarray

```python
f_input = df["Farenheit"].to_numpy()
c_ouput = df["Celcius"].to_numpy()
type(f_input), type(c_ouput)
```

```python
from tensorflow import keras
keras.utils.set_random_seed(42)
```

```python
model = keras.Sequential([
    keras.layers.Dense(units=10, input_shape=[1]),
    keras.layers.Dense(units=1)
])

model.compile(loss="mean_squared_error", optimizer='adam')

model.summary()
```

```python
model.fit(f_input, c_ouput, epochs=100)
```

Dự đoán kết quả

```python
input = np.array([90])
prediction = model.predict(input)
print(input, "degree F is predicted as", prediction[0], "degree C")
```

```python
print("Expected value is", round((input[0] - 32)/1.8, 5), "degree C")
```
